# 10.8 Python Errors

## Syntax Error

compile-time error (translation) that breaks language rules, caught befor program runs. Stops the program from running

## Exception

The syntax is fine (legal), so program runs, but because of this Exception, program will stop and give you an error.

## Logic Error

type of run-time error. program runs normally, gives output as it's supposed to. problem is, if attention isn't being payed, output is wrong. Ex. print(2 + 2) Output: 28 

Logic errors can be difficult to spot. This is why you have to have an understanding of what the output should be

## NOTE: Python is case-sensitive
